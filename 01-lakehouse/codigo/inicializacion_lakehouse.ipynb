## Inicializacion del Entorno Lakehouse
Objetivos del notebook:  
- Crear el catalog principal "dlh_nova_shop".
- Crear el schema para la capa de recursos.
- Crear los schemas para cada capa de trabajo (bronce, plata, oro).
- Crear el schema para la capa de cuarentena.

```python
%sql
CREATE CATALOG IF NOT EXISTS dlh_nova_shop;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS dlh_nova_shop.recursos
COMMENT 'Recursos: Capa de almacenamiento de informacion, simula ser los recursos de datos, en un ejemplo real puede ser una base de datos de SQL server o una base de datos  NOSQL como CosmosDB o DynamoDB, incluso una peticion a una API';
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS dlh_nova_shop.bronce
COMMENT 'Capa Bronce: Ingesta de los datos sin limpieza';
CREATE SCHEMA IF NOT EXISTS dlh_nova_shop.plata
COMMENT 'Capa Plata: Datos limpios y transformados';
CREATE SCHEMA IF NOT EXISTS dlh_nova_shop.oro
COMMENT 'Capa Oro: Datos listos para el negocio';
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS dlh_nova_shop.cuarentena
COMMENT 'Capa Cuarentena: Almacena registros con fallos o inconvenientes';
```
